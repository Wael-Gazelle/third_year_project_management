 4. Basic Feature Engineering


```python

df['text_length'] = df['text'].apply(len)
df['word_count'] = df['text'].apply(lambda x: len(str(x).split()))
df['avg_word_length'] = df['text_length'] / df['word_count']

print("\nSample with new features:")
print(df.head())
```

6. Visualization: Class Comparison

```python
# Exercise 1: ratio of vowels to total letters
def vowel_ratio(text):
    text = text.lower()
    vowels = "aeiou"

    vowel_count = 0
    letter_count = 0

    for char in text:
        if char.isalpha():       # only count letters (skip spaces, punctuation)
            letter_count += 1
            if char in vowels:   # count it as a vowel too
                vowel_count += 1

    return vowel_count / letter_count

# apply the function to every text and save the result in a new column
df['vowel_ratio'] = df['text'].apply(vowel_ratio)

print(df[['text', 'vowel_ratio']].to_string())
```
